```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("income_expense.csv")
x = data["Income"]
y = data["Expense"]

plt.plot(x, y, "o")
plt.xlabel("Income")
plt.ylabel("Expense")
plt.show()
```

```python
print(np.corrcoef(x, y))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import *

data = pd.read_csv("income_expense.csv")
x = data["Income"].values.reshape(-1, 1)
y = data["Expense"].values

model = LinearRegression().fit(x, y)

plt.scatter(x, y)
plt.plot(x, model.predict(x))
plt.xlabel("Income")
plt.ylabel("Expense")
plt.show

print("傾き:", model.coef_)
print("切片:", model.intercept_)
```
