```python
deta = np.array([120, 180, 300, 150])
print(np.sum(deta))
print(np.product(deta))
print(np.square(deta))
print(np.mean(deta))
print(np.min(deta))
print(np.max(deta))
print(np.argmin(deta))
print(np.argmax(deta))
```
