```python
import scipy.stats as st

n = 40
p = 0.5

data_bern = st.bernoulli(p).rvs(size = n)

print(data_bern)
```

```python
import matplotlib.pyplot as plt
plt.hist(data_bern, 2)
```

```python
import scipy.stats as st
n, m, s = 1000, 160, 10
height = st.norm(m, s)

data_normal = height.rvs(n)
print(data_normal)
```

```python
import matplotlib.pyplot as plt
plt.hist(data_normal, 10)
```

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
kde_normal = st.gaussian_kde(data_normal)
plot_x = np.linspace(100, 200, 100)
density_normal = kde_normal(plot_x)

plt.plot(plot_x, density_normal)
plt.show()
```

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np

kde_normal = st.gaussian_kde(data_normal)
plot_x = np.linspace(100, 200, 100)
density_normal = kde_normal(plot_x)

fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.plot(plot_x, density_normal)
ax1.set_ylabel("Density function")

ax2.hist(data_normal, bins = 10, alpha = 0.5, edgecolor = "black")
ax2.set_ylabel("Histogram")

plt.show()
```

```python
import statistics
print(statistics.mean(data_normal))
print(statistics.median(data_normal))
print(statistics.mode(data_normal))
print(statistics.pvariance(data_normal))
print(statistics.pstdev(data_normal))
```

```python
import scipy.stats as st

n = 40
lam = 3.0

data_poisson = st.poisson(mu = lam).rvs(size = n)

print(data_poisson)
```

```python
import matplotlib.pyplot as plt
plt.hist(data_poisson, 10)
```

```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np

kde_poisson = st.gaussian_kde(data_poisson)
plot_x = np.linspace(0, 10, 200)
density_poisson = kde_poisson(plot_x)

fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.plot(plot_x, density_poisson)
ax1.set_ylabel("Density function")

ax2.hist(data_poisson, bins = 10, alpha = 0.75, edgecolor = "black")
ax2.set_ylabel("Histogram")

plt.show()
```
