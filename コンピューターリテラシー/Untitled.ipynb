```python
import cvxpy as cp
x, y = cp.Variable(), cp.Variable()
obj = cp.Maximize(2*x + y)
cons = [(x - 1)**2 - y <= 0, 
        -3*x + y <= 7]
P = cp.Problem(obj, cons)
P.solve()
print(P.value, x.value, y.value)
```
