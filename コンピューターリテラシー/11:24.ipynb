```python
A = np.array([[6.0, 2.0], [2.0, 5.0]])
rankA = np.linalg.matrix_rank(A)

print(rankA)
```

```python
B = np.array([[6.0, 2.0, 4.0], [2.0, 5.0, 2.0], [3.0, 5.0, 1.0]])
rankB = np.linalg.matrix_rank(B)

print(rankB)
```

```python
A = np.array([[6.0, 2.0], [2.0, 5.0]])

invA = np.linalg.inv(A)

print(invA)
```

```python
A = np.array([[6.0, 2.0], [2.0, 5.0]])
B = np.array([[6.0, 3.0], [2.0, 1.0]])

detA = np.linalg.det(A)
detB = np.linalg.det(B)

print(detA)
print(detB)
```

```python
A = np.array([[6.0, 2.0], [2.0, 5.0]])
B = np.array([[6.0, 3.0], [2.0, 1.0]])

detA = np.linalg.det(A)
detB = np.linalg.det(B)

print(f"Determinant A : {detA: 1f}")
print(f"Determinant B : {detB: 1f}")
```

```python
A = np.array([[3.0, 2.0], [4.0, 1.0]])

E, V = np.linalg.eig(A)

print(f"Eigenvalue :\n {E}")
print(f"Eigenvector :\n {V}")
```
